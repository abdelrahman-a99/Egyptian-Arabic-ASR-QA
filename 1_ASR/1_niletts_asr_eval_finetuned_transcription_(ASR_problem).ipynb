```python
from torch.utils.data import Dataset
import soundfile as sf
import numpy as np
import torch
from scipy.signal import resample

class WhisperDataset(Dataset):

    def __init__(self, dataframe, processor):
        self.dataframe = dataframe.reset_index(drop=True)
        self.processor = processor

    def __len__(self):
        return len(self.dataframe)

    def __getitem__(self, idx):

        row = self.dataframe.iloc[idx]

        audio_path = row["audio_path"]
        text = row["text"]

        # LOAD AUDIO USING SOUNDFILE
        audio_array, sampling_rate = sf.read(audio_path)

        # stereo -> mono
        if len(audio_array.shape) > 1:
            audio_array = np.mean(audio_array, axis=1)

        # RESAMPLE TO 16k
        if sampling_rate != 16000:
            num_samples = int(len(audio_array) * 16000 / sampling_rate)

            audio_array = resample(
            audio_array,
            num_samples
            )
            sampling_rate = 16000

        # PROCESS INPUTS
        input_features = self.processor.feature_extractor(
            audio_array,
            sampling_rate=sampling_rate
        ).input_features[0]

        # PROCESS LABELS
        labels = self.processor.tokenizer(
            text,
            truncation=True,
            max_length=448
        ).input_ids

        return {
            "input_features": input_features,
            "labels": labels
        }
```

```python
import torch
from dataclasses import dataclass
from typing import Any, Dict, List, Union

@dataclass
class DataCollatorSpeechSeq2SeqWithPadding:
    processor: Any
    decoder_start_token_id: int

    def __call__(
        self,
        features: List[Dict[str, Union[List[int], torch.Tensor]]]
    ) -> Dict[str, torch.Tensor]:

        input_features = [
            {"input_features": feature["input_features"]}
            for feature in features
        ]

        batch = self.processor.feature_extractor.pad(
            input_features,
            return_tensors="pt"
        )

        label_features = [
            {"input_ids": feature["labels"]}
            for feature in features
        ]

        labels_batch = self.processor.tokenizer.pad(
            label_features,
            return_tensors="pt"
        )

        labels = labels_batch["input_ids"].masked_fill(
            labels_batch.attention_mask.ne(1),
            -100
        )

        if (labels[:, 0] == self.decoder_start_token_id).all().cpu().item():
            labels = labels[:, 1:]

        batch["labels"] = labels

        return batch


data_collator = DataCollatorSpeechSeq2SeqWithPadding(
    processor=processor,
    decoder_start_token_id=model.config.decoder_start_token_id
)
```

```python
from jiwer import wer, cer

def compute_metrics(pred):

    pred_ids = pred.predictions
    label_ids = pred.label_ids

    label_ids[label_ids == -100] = processor.tokenizer.pad_token_id

    pred_str = processor.tokenizer.batch_decode(
        pred_ids,
        skip_special_tokens=True
    )

    label_str = processor.tokenizer.batch_decode(
        label_ids,
        skip_special_tokens=True
    )

    return {
        "wer": wer(label_str, pred_str),
        "cer": cer(label_str, pred_str),
    }
```

```python
import soundfile as sf
import numpy as np
import pandas as pd
import torch
import time

from tqdm.auto import tqdm
from scipy.signal import resample
from jiwer import wer, cer

asr_predictions = []
wer_scores = []
cer_scores = []

start_time = time.time()

total_samples = len(train_df)

for idx in tqdm(range(total_samples), desc="Generating ASR"):

    try:
        row = train_df.iloc[idx]

        audio_path = row["audio_path"]

        reference_text = str(row["text"])

        audio_array, sampling_rate = sf.read(audio_path)

        # stereo -> mono
        if len(audio_array.shape) > 1:
            audio_array = np.mean(audio_array, axis=1)

        # convert to float32
        audio_array = audio_array.astype(np.float32)

        # RESAMPLE TO 16K
        if sampling_rate != 16000:

            num_samples = int(
                len(audio_array) * 16000 / sampling_rate
            )

            audio_array = resample(
                audio_array,
                num_samples
            ).astype(np.float32)

        # PROCESS INPUT
        inputs = processor(
            audio_array,
            sampling_rate=16000,
            return_tensors="pt"
        )

        input_features = inputs.input_features.to(device)

        # GENERATE TRANSCRIPT
        with torch.no_grad():

            predicted_ids = model.generate(
                input_features
            )

        predicted_text = processor.batch_decode(
            predicted_ids,
            skip_special_tokens=True
        )[0]

        # CALCULATE WER/CER
        sample_wer = wer(
            reference_text,
            predicted_text
        )

        sample_cer = cer(
            reference_text,
            predicted_text
        )

        # SAVE RESULTS
        asr_predictions.append(predicted_text)
        wer_scores.append(sample_wer)
        cer_scores.append(sample_cer)

        # -----------------------------
        # LOGGING

        if idx % 50 == 0:
            elapsed_time = time.time() - start_time
            avg_time_per_sample = elapsed_time / (idx + 1)
            remaining_samples = total_samples - (idx + 1)
            eta_seconds = remaining_samples * avg_time_per_sample
            eta_minutes = eta_seconds / 60
            running_wer = np.mean(wer_scores)
            running_cer = np.mean(cer_scores)
            progress_percent = ((idx + 1) / total_samples) * 100

            print(f"Processed: {idx+1}/{total_samples}")
            print(f"Progress: {progress_percent:.2f}%")
            print(f"Elapsed: {elapsed_time/60:.2f} min")
            print(f"ETA Remaining: {eta_minutes:.2f} min")
            print(f"Running Avg WER: {running_wer:.4f}")
            print(f"Running Avg CER: {running_cer:.4f}")
            print("\nREFERENCE:")
            print(reference_text[:200])
            print("\nPREDICTION:")
            print(predicted_text[:200])
            print(f"Sample WER: {sample_wer:.4f}")
            print(f"Sample CER: {sample_cer:.4f}")

    except Exception as e:
        print(f"\nERROR AT INDEX {idx}: {e}")

        asr_predictions.append("")
        wer_scores.append(1.0)
        cer_scores.append(1.0)


train_df["asr_text"] = asr_predictions
train_df["wer"] = wer_scores
train_df["cer"] = cer_scores

overall_wer = np.mean(wer_scores)
overall_cer = np.mean(cer_scores)


print("FINAL RESULTS")
print(f"Overall WER: {overall_wer:.4f}")
print(f"Overall CER: {overall_cer:.4f}")

# =====
# OUTPUT_CSV = "niletts_asr_results.csv"
OUTPUT_CSV = ASR_DIR / "niletts_asr_results.csv"
# =====

train_df.to_csv(
    OUTPUT_CSV,
    index=False
)

print(f"\nSaved file: {OUTPUT_CSV}")

train_df.head()
```

```python
class NileTTSWhisperDataset(TorchDataset):
    def __init__(self, dataframe, processor, sampling_rate=16000):
        self.df = dataframe.reset_index(drop=True)
        self.processor = processor
        self.sampling_rate = sampling_rate

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]

        audio_path = row["audio_path"]
        text = row["text"]

        audio_array, sr = librosa.load(
            audio_path,
            sr=self.sampling_rate,
            mono=True
        )

        input_features = self.processor.feature_extractor(
            audio_array,
            sampling_rate=self.sampling_rate
        ).input_features[0]

        # =====
        # labels = self.processor.tokenizer(text).input_ids
        labels = self.processor.tokenizer(
            text,
            truncation=True,
            max_length=448
        ).input_ids
        # =====

        return {
            "input_features": input_features,
            "labels": labels
        }


train_dataset = NileTTSWhisperDataset(train_df, processor)
eval_dataset = NileTTSWhisperDataset(eval_df, processor)

print("Train dataset:", len(train_dataset))
print("Eval dataset:", len(eval_dataset))
```

```python
@dataclass
class DataCollatorSpeechSeq2SeqWithPadding:
    processor: Any
    decoder_start_token_id: int

    def __call__(
        self,
        features: List[Dict[str, Union[List[int], torch.Tensor]]]
    ) -> Dict[str, torch.Tensor]:

        input_features = [
            {"input_features": feature["input_features"]}
            for feature in features
        ]

        batch = self.processor.feature_extractor.pad(
            input_features,
            return_tensors="pt"
        )

        label_features = [
            {"input_ids": feature["labels"]}
            for feature in features
        ]

        labels_batch = self.processor.tokenizer.pad(
            label_features,
            return_tensors="pt"
        )

        labels = labels_batch["input_ids"].masked_fill(
            labels_batch.attention_mask.ne(1),
            -100
        )

        if (labels[:, 0] == self.decoder_start_token_id).all().cpu().item():
            labels = labels[:, 1:]

        batch["labels"] = labels

        return batch


data_collator = DataCollatorSpeechSeq2SeqWithPadding(
    processor=processor,
    decoder_start_token_id=model.config.decoder_start_token_id
)

print("Data collator ready.")
```

```python
def prepare_dataset(batch):
    audio = batch["audio"]

    # Extract features inline — no intermediate variable held in memory
    batch["input_features"] = processor.feature_extractor(
        audio["array"],
        sampling_rate=audio["sampling_rate"],
        return_tensors="np"          # stay on CPU / numpy, not pt
    ).input_features[0]

    # Tokenise text

    # =====
    # batch["labels"] = processor.tokenizer(batch["text"]).input_ids
    batch["labels"] = processor.tokenizer(
        batch["text"],
        truncation=True,
        max_length=448
    ).input_ids
    # =====

    # Explicitly delete the raw audio from the batch to free RAM immediately
    del audio
    return batch
```

```python
from dataclasses import dataclass
from typing import Any, Dict, List, Union
import torch

@dataclass
class DataCollatorSpeechSeq2SeqWithPadding:
    processor: Any
    decoder_start_token_id: int

    def __call__(
        self,
        features: List[Dict[str, Union[List[int], torch.Tensor]]]
    ) -> Dict[str, torch.Tensor]:

        input_features = [
            {"input_features": feature["input_features"]}
            for feature in features
        ]
        batch = self.processor.feature_extractor.pad(
            input_features,
            return_tensors="pt"
        )

        label_features = [
            {"input_ids": feature["labels"]}
            for feature in features
        ]
        labels_batch = self.processor.tokenizer.pad(
            label_features,
            return_tensors="pt"
        )

        labels = labels_batch["input_ids"].masked_fill(
            labels_batch.attention_mask.ne(1), -100
        )

        # ✅ FIX: Delete labels_batch immediately — the padded input_ids tensor
        # and attention_mask are no longer needed once we've applied the mask.
        del labels_batch

        if (labels[:, 0] == self.decoder_start_token_id).all().cpu().item():
            labels = labels[:, 1:]

        batch["labels"] = labels
        return batch


data_collator = DataCollatorSpeechSeq2SeqWithPadding(
    processor=processor,
    decoder_start_token_id=model.config.decoder_start_token_id
)
```

```python
import re

def normalize_arabic_for_eval(text):
    text = str(text)
    text = re.sub(r"[\u064B-\u065F\u0670]", "", text)   # remove tashkeel
    text = re.sub(r"[إأآا]", "ا", text)
    text = re.sub(r"ى", "ي", text)
    text = re.sub(r"ة", "ه", text)
    text = re.sub(r"[^\w\s\u0600-\u06FF]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text


def compute_metrics(pred):
    pred_ids  = pred.predictions
    label_ids = pred.label_ids

    label_ids[label_ids == -100] = processor.tokenizer.pad_token_id

    pred_str  = processor.tokenizer.batch_decode(pred_ids,  skip_special_tokens=True)
    label_str = processor.tokenizer.batch_decode(label_ids, skip_special_tokens=True)

    # ✅ FIX: Delete the raw id arrays right after decoding — they can be
    # hundreds of MB (batch_size × sequence_length × int64) and are never
    # needed again once we have the decoded strings.
    del pred_ids, label_ids

    pred_str  = [normalize_arabic_for_eval(x) for x in pred_str]
    label_str = [normalize_arabic_for_eval(x) for x in label_str]

    wer_score = wer_metric.compute(predictions=pred_str, references=label_str)
    cer_score = cer_metric.compute(predictions=pred_str, references=label_str)

    # ✅ FIX: Delete decoded string lists after metric computation
    del pred_str, label_str

    return {"wer": wer_score, "cer": cer_score}
```

```python
import whisper

# ✅ FIX: Load models ONCE here — the original code reloaded the model
# on every single audio file, wasting GBs of RAM and minutes of time.
_whisper_cache = {}

def get_whisper_model(model_name):
    """Return a cached model; load only on first call."""
    if model_name not in _whisper_cache:
        _whisper_cache[model_name] = whisper.load_model(model_name)
    return _whisper_cache[model_name]

def transcribe_audio(model_name, audio_path):
    model = get_whisper_model(model_name)
    result = model.transcribe(audio_path)
    return result['text']
```

```python
import torchaudio
from transformers import WhisperProcessor, WhisperForConditionalGeneration
import torch
import os
from tqdm import tqdm

fine_tuned_model_name = "MAdel121/whisper-small-egyptian-arabic"
processor_ft = WhisperProcessor.from_pretrained(fine_tuned_model_name)
model_ft     = WhisperForConditionalGeneration.from_pretrained(fine_tuned_model_name)

device = "cuda" if torch.cuda.is_available() else "cpu"
model_ft.to(device)

# ✅ FIX: removed per-sample print() calls — printing inside a loop that runs
# thousands of times is surprisingly slow and floods the notebook output.
def transcribe_finetuned_audio(audio_path):
    waveform, sample_rate = torchaudio.load(audio_path)

    if sample_rate != 16000:
        waveform = torchaudio.transforms.Resample(sample_rate, 16000)(waveform)

    # Average channels → mono
    if waveform.shape[0] > 1:
        waveform = waveform.mean(dim=0)
    else:
        waveform = waveform.squeeze(0)

    audio_input = processor_ft(
        waveform.numpy(), return_tensors="pt", sampling_rate=16000
    ).input_features.to(device)

    with torch.no_grad():                      # ✅ no_grad saves GPU memory
        predicted_ids = model_ft.generate(audio_input)

    return processor_ft.decode(predicted_ids[0], skip_special_tokens=True)


# Quick sanity check on one file
test_audio_path = eval_df['audio_path'].iloc[0]
print("Test transcription:", transcribe_finetuned_audio(test_audio_path))
```
